## <u>**MODELO FINAL**</u>


Cargamos el modelo final creado en C1 y hacemos las predicciones

```python
import pandas as pd
import numpy as np
import joblib

# Cargamos el modelo final
modelo_final = joblib.load('modelo_final.joblib')

# Cargamos los datos de competición
df_comp = pd.read_pickle('bank_competition.pkl')

# Mismo preproceso de pdays que en C1
df_comp['contacted_before'] = (df_comp['pdays'] != -1).astype(int)
df_comp['pdays'] = df_comp['pdays'].replace(-1, np.nan)

# Hacemos predicciones
predicciones = modelo_final.predict(df_comp)

# Guardamos las predicciones
pd.DataFrame({'deposit': predicciones}).to_csv('predicciones.csv', index=False)
print(pd.Series(predicciones).value_counts())
```
